# Technoeconomic Plotter
This notebook compares cost over time for Lab CFPP, Lab CBPP, and Lab CFPP (with micropad recycling).

```python
import matplotlib.pyplot as plt
import pandas as pd
```

# Lab Calculations

```python
# Load data
lab_cfpp = pd.read_csv(r'./data/raw/CFPP Lab.csv')
lab_cbpp = pd.read_csv(r'./data/raw/CBPP Lab.csv')
lab_cfpp_recycled = pd.read_csv(r'./data/raw/CFPP Lab (recycle mpads).csv')

# column names [time (hours),protein (mg),cost ($)]
```

```python
# for each dataset, add a column of cost in ($ millions)
lab_cbpp['cost ($ millions)'] = lab_cbpp['cost ($)'] / 1_000_000
lab_cfpp['cost ($ millions)'] = lab_cfpp['cost ($)'] / 1_000_000
lab_cfpp_recycled['cost ($ millions)'] = lab_cfpp_recycled['cost ($)'] / 1_000_000
```

```python
# for each dataset add column of cost efficiency for cost/mg at timepoint
lab_cbpp['cost efficiency ($/mg)'] = lab_cbpp['cost ($)'] / lab_cbpp['protein (mg)']
lab_cfpp['cost efficiency ($/mg)'] = lab_cfpp['cost ($)'] / lab_cfpp['protein (mg)']
lab_cfpp_recycled['cost efficiency ($/mg)'] = lab_cfpp_recycled['cost ($)'] / lab_cfpp_recycled['protein (mg)']

lab_cfpp.head()
```

```python
# only keep rows where time (hours) < 200
lab_cbpp = lab_cbpp[lab_cbpp['time (hours)'] < 200]
lab_cfpp = lab_cfpp[lab_cfpp['time (hours)'] < 200]
lab_cfpp_recycled = lab_cfpp_recycled[lab_cfpp_recycled['time (hours)'] < 200]
```

```python
# Combined plots in one row
fig, axs = plt.subplots(1, 3, figsize=(18, 5))

# Plot 1: Cost ($ millions) over time
axs[0].plot(lab_cbpp['time (hours)'], lab_cbpp['cost ($ millions)'], label='Industrial Scale CBPP', color='green')
axs[0].plot(lab_cfpp_recycled['time (hours)'], lab_cfpp_recycled['cost ($ millions)'], label='Industrial Scale CFPP (Recycled)', color='orange')
axs[0].plot(lab_cfpp['time (hours)'], lab_cfpp['cost ($ millions)'], label='Industrial Scale CFPP', color='blue')
# Add dotted vertical line at hour 40
axs[0].axvline(x=40, color='gray', linestyle=':', linewidth=2, label='One Cycle, producing 200g')
axs[0].set_xlabel('Time (hours)')
axs[0].set_ylabel('Cost ($ millions)')
axs[0].set_title('Total Cost Over Time')
axs[0].legend()
axs[0].grid(True)

# Plot 2: Protein (mg) over time
axs[1].plot(lab_cbpp['time (hours)'], lab_cbpp['protein (mg)'], label='Industrial Scale CBPP', color='green')
axs[1].plot(lab_cfpp_recycled['time (hours)'], lab_cfpp_recycled['protein (mg)'], label='Industrial Scale CFPP (Recycled)', color='blue')
axs[1].plot(lab_cfpp['time (hours)'], lab_cfpp['protein (mg)'], label='Industrial Scale CFPP (all types)', color='blue')
axs[1].set_xlabel('Time (hours)')
axs[1].set_ylabel('Protein (mg)')
axs[1].set_title('Protein Over Time')
axs[1].legend()
axs[1].grid(True)

# Plot 3: Cost efficiency ($/mg) over time
axs[2].plot(lab_cbpp['time (hours)'], lab_cbpp['cost efficiency ($/mg)'], label='CBPP Industrial Scale', color='green')
axs[2].plot(lab_cfpp['time (hours)'], lab_cfpp['cost efficiency ($/mg)'], label='CFPP Industrial Scale', color='blue')
axs[2].plot(lab_cfpp_recycled['time (hours)'], lab_cfpp_recycled['cost efficiency ($/mg)'], label='CFPP Industrial Scale Recycled', color='orange')
axs[2].set_xlabel('Time (hours)')
axs[2].set_ylabel('Cost Efficiency ($/mg)')
axs[2].set_title('Cost Efficiency Over Time')
axs[2].legend()
axs[2].grid(True)

plt.tight_layout()

plt.subplots_adjust(top=0.88)
plt.suptitle('Industrial Scale Comparisons')
plt.show()
```

# Breakdown Costs for CFPP

```python
cost_breakdown = pd.read_csv(r'./data/final_costs_breakdown.csv')
```

```python
cfpp_lab = cost_breakdown[cost_breakdown['name'] == 'CFPP Lab']
cfpp_ind = cost_breakdown[cost_breakdown['name'] == 'CFPP Industry']
cfpp_proto = cost_breakdown[cost_breakdown['name'] == 'CFPP Prototype'] 
```

```python
# for each dataframe, create a pie chart of specific items contributing to total cost
def pie_labels(values, labels):
    total = sum(values)
    return [f'{label}: {value/total*100:.1f}%' for label, value in zip(labels, values)]

fig, axs = plt.subplots(1, 3, figsize=(18, 5))

# Pie chart for CFPP Lab
lab_labels = pie_labels(cfpp_lab['cost_$'], cfpp_lab['specific item'])
axs[0].pie(cfpp_lab['cost_$'], labels=None, autopct='%1.1f%%', startangle=140)
axs[0].set_title('CFPP Lab Cost Breakdown')
axs[0].legend(lab_labels, loc='lower right')

# Pie chart for CFPP Industry
ind_labels = pie_labels(cfpp_ind['cost_$'], cfpp_ind['specific item'])
axs[1].pie(cfpp_ind['cost_$'], labels=None, autopct='%1.1f%%', startangle=140)
axs[1].set_title('CFPP Industry Cost Breakdown')
axs[1].legend(ind_labels, loc='lower right')

# Pie chart for CFPP Prototype
proto_labels = pie_labels(cfpp_proto['cost_$'], cfpp_proto['specific item'])
axs[2].pie(cfpp_proto['cost_$'], labels=None, autopct='%1.1f%%', startangle=140)
axs[2].set_title('CFPP Prototype Cost Breakdown')
axs[2].legend(proto_labels, loc='lower right')

plt.tight_layout()
plt.subplots_adjust(top=0.88)
plt.suptitle('Cost Breakdown by Specific Item for CFPP at Different Scales')
plt.show()
```
